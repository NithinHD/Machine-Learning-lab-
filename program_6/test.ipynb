```python
import pandas as pd
```

```python
df=pd.read_csv('smaller_adult.csv')
df=pd.DataFrame(df)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.drop_duplicates()
```

```python
y=df['income']
x=df.drop(['income'],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
y=LabelEncoder().fit_transform(y)
y=pd.DataFrame(y)
y
```

```python
#tak
numerical_features=x.select_dtypes('number')
catagorical_features=x.select_dtypes('object')
numerical_features
```

```python
converted_catagorical_features=pd.get_dummies(catagorical_features)
all_features=[converted_catagorical_features,numerical_features]
newX=pd.concat(all_features,axis=1,join='inner')
converted_catagorical_features.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(newX,y,test_size=0.3)
```

```python
from sklearn.svm import SVC
model=SVC(kernel='linear',gamma='auto').fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
