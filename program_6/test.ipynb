```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Smaller_adult.csv')
df=pd.DataFrame(df)
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.isin(['?']).sum()
```

```python
df=df.replace('?',np.NaN)
df.dropna(how='any',inplace=True)
```

```python
y=df['income']
x=df.drop('income',axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
y=LabelEncoder().fit_transform(y)
y=pd.DataFrame(y)
y
```

```python
numerical_features=x.select_dtypes('number')
catatorical_feature=x.select_dtypes('object')
```

```python
catatorical_feature=pd.get_dummies(catatorical_feature)
x=pd.concat([catatorical_feature,numerical_features],axis=1,join='inner')
x.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,np.ravel(y),test_size=0.3)
```

```python
from sklearn.svm import  SVC
model=SVC(kernel='linear').fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import  accuracy_score
accuracy_score(y_test,y_pred)
```
