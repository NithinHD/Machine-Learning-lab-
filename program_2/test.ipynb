```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
rd=pd.read_csv('wine-clustering.csv')
df=pd.DataFrame(rd)
```

```python
#description of the data
df.describe().T
```

```python
#info of the data
df.info()
```

```python
sns.pairplot(df)
```

```python
from sklearn.cluster import KMeans
#colums to be selected
selected_features=df[['OD280','Alcohol']]
#random state is used to get the same result every time
kmeans_obj=KMeans(n_clusters=3,random_state=42)
#fitting the data
kmeans_obj.fit(selected_features)
#predicting cluster labels for the data
y_kmeans=kmeans_obj.fit_predict(selected_features)
print(y_kmeans)
```

```python
#printing the cluster centers  
centers=kmeans_obj.cluster_centers_
centers
```

```python
import matplotlib.pyplot as plt
#visualizing the clusters craeted
sns.scatterplot(x=selected_features['OD280'],y=selected_features['Alcohol'],hue=kmeans_obj.labels_)
plt.scatter(kmeans_obj.cluster_centers_[:, 0],kmeans_obj.cluster_centers_[:, 1],c='red',s=200)
```
